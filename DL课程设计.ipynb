```python
"""
5.1 基准模型（Baseline）
尝试使用简单的全连接神经网络（MLP）作为基准，但发现其无法有效提取图像的空间特征，且参数量过大，容易过拟合，准确率较低。
5.2 最终模型架构
网络结构设计：构建了一个轻量级的卷积神经网络，具体架构如下：
卷积层组 1：Conv2D (32 filters, 3x3 kernel) + ReLU + MaxPooling
卷积层组 2：Conv2D (64 filters, 3x3 kernel) + ReLU + MaxPooling
卷积层组 3：Conv2D (128 filters, 3x3 kernel) + ReLU + MaxPooling
展平层：Flatten
全连接层 1：Dense (512 units) + ReLU + Dropout (0.5)
输出层：Dense (1 unit) + Sigmoid
理论依据：卷积层用于提取局部特征（如边缘、纹理），池化层用于降维和保留主要特征，全连接层用于整合特征进行分类。Dropout层用于随机丢弃神经元，进一步提升模型泛化能力。


"""
```

```python
"""
六、实验与结果分析
6.1 实验环境
硬件：CPU: Intel Core i7 / GPU: NVIDIA GTX 1660 Ti (或根据你实际情况填写)
软件：Python 3.8, TensorFlow/Keras 或 PyTorch, OpenCV, NumPy, Matplotlib
6.2 评价指标
准确率：预测正确的样本数占总样本数的比例。
损失值：二元交叉熵损失，用于衡量预测概率分布与真实标签的差异。
6.3 超参数设置与调优
优化器：Adam (自适应学习率优化算法)。
学习率：初始学习率设置为 0.0001。
批大小：Batch Size = 32。
迭代次数：Epochs = 20（采用了早停法防止过拟合）。


"""
```

```python
# 导入必要的库
import os
import numpy as np
import matplotlib.pyplot as plt

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

# 设置matplotlib显示中文和负号（如果需要）
plt.rcParams['font.sans-serif'] = ['SimHei'] 
plt.rcParams['axes.unicode_minus'] = False

print("TensorFlow 版本:", tf.__version__)
```

```python
# --- 配置参数 ---
# TODO: 请将此路径修改为你本地数据集的根目录
# 你的数据集目录结构应如下所示：
# /path/to/your/dataset/
# ├── train/
# │   ├── cats/
# │   └── dogs/
# └── validation/
#     ├── cats/
#     └── dogs/
base_dir = '/path/to/your/dataset' 

train_dir = os.path.join(base_dir, 'train')
validation_dir = os.path.join(base_dir, 'validation')

# 图像参数
img_width, img_height = 150, 150
batch_size = 32

# --- 数据增强与归一化 ---
# 训练集：进行数据增强和归一化
train_datagen = ImageDataGenerator(
    rescale=1./255,          # 归一化
    rotation_range=40,       # 随机旋转
    width_shift_range=0.2,   # 水平平移
    height_shift_range=0.2,  # 垂直平移
    shear_range=0.2,         # 剪切变换
    zoom_range=0.2,          # 随机缩放
    horizontal_flip=True,    # 随机水平翻转
    fill_mode='nearest'
)

# 验证集：仅进行归一化
validation_datagen = ImageDataGenerator(rescale=1./255)

# --- 从目录加载数据 ---
print("正在加载训练数据...")
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(img_width, img_height),
    batch_size=batch_size,
    class_mode='binary' # 二分类
)

print("正在加载验证数据...")
validation_generator = validation_datagen.flow_from_directory(
    validation_dir,
    target_size=(img_width, img_height),
    batch_size=batch_size,
    class_mode='binary'
)

# 获取类别标签
class_labels = list(train_generator.class_indices.keys())
print(f"类别标签: {class_labels}")
```

```python
# --- 构建CNN模型 ---
model = models.Sequential([
    # 第1个卷积块
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(img_width, img_height, 3)),
    layers.MaxPooling2D(2, 2),
    
    # 第2个卷积块
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D(2, 2),
    
    # 第3个卷积块
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D(2, 2),
    
    # 展平并连接全连接层
    layers.Flatten(),
    layers.Dropout(0.5), # 防止过拟合
    layers.Dense(512, activation='relu'),
    layers.Dense(1, activation='sigmoid') # 二分类输出
])

# 打印模型结构
model.summary()

# --- 编译模型 ---
model.compile(
    loss='binary_crossentropy',
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    metrics=['accuracy']
)
```

```python
# --- 设置训练参数 ---
epochs = 20

# 设置回调函数，例如早停法，防止过拟合
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

# --- 开始训练 ---
print("开始训练模型...")
history = model.fit(
    train_generator,
    steps_per_epoch=train_generator.samples // batch_size,
    epochs=epochs,
    validation_data=validation_generator,
    validation_steps=validation_generator.samples // batch_size,
    callbacks=[early_stopping],
    verbose=1
)

# 保存模型（可选）
# model.save('cats_vs_dogs_cnn.h5')
```

```python
# --- 绘制训练过程中的准确率和损失曲线 ---
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']

epochs_range = range(len(acc))

plt.figure(figsize=(12, 4))

# 准确率曲线
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.legend(loc='lower right')
plt.title('训练与验证准确率')
plt.grid(True)

# 损失曲线
plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.legend(loc='upper right')
plt.title('训练与验证损失')
plt.grid(True)

plt.show()
```

```python
# --- 在验证集上进行预测 ---
print("正在评估模型...")
validation_generator.reset()
pred = model.predict(validation_generator, steps=np.ceil(validation_generator.samples / batch_size))
predicted_classes = (pred > 0.5).astype(int).flatten() # 概率转类别 (0或1)
true_classes = validation_generator.classes

# --- 计算混淆矩阵 ---
cm = confusion_matrix(true_classes, predicted_classes)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_labels, yticklabels=class_labels)
plt.ylabel('真实标签')
plt.xlabel('预测标签')
plt.title('混淆矩阵')
plt.show()

# --- 打印分类报告 ---
print("\n分类报告:")
print(classification_report(true_classes, predicted_classes, target_names=class_labels))

# --- 打印最终准确率 ---
final_val_acc = history.history['val_accuracy'][-1]
print(f"\n模型在验证集上的最终准确率约为: {final_val_acc*100:.2f}%")
```
